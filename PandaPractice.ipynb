9️⃣ Filtering Data (VERY IMPORTANT)

```python
df[df['Age'] > 25]
```

🔹 Multiple Conditions

```python
df[(df['Age'] > 25) & (df['Salary' ] > 50000)]
```

🔟 Modifying Data

```python
df['tax']=df['Salary'] *0.1
```

Update Values

```python
df.loc[df['Age'] > 25, 'Salary'] += 5000
```

1️⃣3️⃣ GroupBy (MOST POWERFUL FEATURE)

Think:

“Split → Apply → Combine”

```python
df.groupby('Country')['Income'].mean()
```

```python
df.groupby('Country').agg({'Income':['mean','max'],'Age':'mean'})
```
